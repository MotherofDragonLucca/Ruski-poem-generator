```python
import requests
from rich import print
from rich.markdown import Markdown

def display_current_weather(location):
    """Get the real-time current temperature and condition in a location"""
    api_key = "2cdb3f5e83a43fbc4a0edc02t2ef5ofc"
    api_url = f"https://api.shecodes.io/weather/v1/current?query={location}&key={api_key}&units=metric"

    response = requests.get(api_url)
    response_data = response.json()

    if "temperature" in response_data and "condition" in response_data:
        temperature = round(response_data["temperature"]["current"])
        condition = response_data["condition"]["description"]
        print(f"The current temperature in [bold]{location}[/bold] is [bold]{temperature}°C[/bold], {condition}.")
    else:
        print(f"Ooops, could not retrieve weather data for {location}. Please try again.")
        print(f"API Response: {response_data}")

def generate_itinerary(origin, destination, duration):
    """Generate a travel itinerary between 2 places using AI"""
    api_key = "2cdb3f5e83a43fbc4a0edc02t2ef5ofc"
    print(f"\n\nGenerating itinerary from {origin} to {destination}...\n")

    prompt = (
        f"Create a fun, efficient travel itinerary from {origin} to {destination} over {duration} days. "
        "Keep it concise—no more than 50 lines. Summarize days into weeks if running out of space. "
        "Use fast, practical transportation options (e.g., trains, flights, buses). For each day or stop, include:\n"
        "- Approximate daily cost in local currency 💰\n"
        "- Weather conditions ☀️🌧️❄️\n"
        "- Packing suggestions based on weather (e.g., umbrella ☂️, sun hat 👒, jacket 🧥)\n"
        "- One must-visit attraction 🗺️\n"
        "- One recommended restaurant or café 🍽️\n"
        "Add a touch of humor and sprinkle in emojis sparingly for flair."
    )

    context = (
        "You are a witty and intuitive travel planner who crafts personalized, adventurous itineraries across Earth. "
        "You understand the importance of efficient travel, local experiences, and weather-based packing tips. "
        "Your tone is friendly and humorous, and you use emojis tastefully to enhance the experience. "
        "You recommend realistic transportation options and highlight cultural gems, local cuisine, and scenic spots."
    )

    api_url = f"https://api.shecodes.io/ai/v1/generate?prompt={prompt}&context={context}&key={api_key}"
    response = requests.get(api_url)
    response_data = response.json()

    if "answer" in response_data:
        itinerary = Markdown(response_data["answer"])
        print(itinerary)
    else:
        print("Oops, could not generate itinerary, please try again.")
        print(f"API Response: {response_data}")

def welcome():
    print("\n\n🌍[bold white]Welcome to the Earth Explorer Travel Planner![/bold white]🌍\n")

def credit():
    print("The Earth Explorer Travel Planner was built by [bold]<a href="https://github.com/MotherofDragonLucca" target="_blank">Farida
                Scandrett</a> and is <a href="https://github.com/MotherofDragonLucca/Ruski-poem-generator"
                target="_blank">open-sourced on GitHub</a>[/bold], safe travels.✈️[/white]")


# Call welcome function
welcome()

# Get user input
origin = input("What’s your home location? ").strip().capitalize()
destination = input("What's your desired destination? ").strip().capitalize()
duration = input("How long is your trip? (Enter number of days): ")

if origin and destination and duration.isdigit() and int(duration) > 0:
    display_current_weather(origin)
    display_current_weather(destination)
    generate_itinerary(origin, destination, duration)
    credit()
else:
    print("Oops, please try again.")
```
