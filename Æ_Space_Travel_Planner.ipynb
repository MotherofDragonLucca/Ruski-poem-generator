def generate_itinerary(origin, destination, duration):
    """Generate a travel itinerary between 2 places using AI"""
    print(f"\n\nGenerating itinerary from {origin} to {destination}...\n")
    prompt = (
        f"Generate a travel itinerary from {origin} to {destination} in {duration} days. "
        "Use direct route with efficient transportation suitable for space travel, keep it short no more than 30 lines, the if required summarise days into groups of 2 to 3 day or weeks to fit within these 30 lines and always assume it's a return trip. Check form of transportation is available to public, no space travels training required. Keep it affordable."
        "Include emojis but use them sparingly. Add approximate price for each day in local currency and weather conditions of each step, then based on the weather condition suggest to pack such as umbrella, sun hat, jacket emoji. Include one restaurant or cafe and one attraction in each location to visit."
    )

    context = (
        "You're a skilled space travel agent who intuitively anticipates users' needs and "
        "recommends intrepid travel recommendations highly tailored to their needs, use humour."
        "suggesting fast and efficient forms of transportation."
    )

    api_url = f"https://api.shecodes.io/ai/v1/generate?prompt={prompt}&context={context}&key={API_KEY}"
    response = requests.get(api_url)
    response_data = response.json()

    if "answer" in response_data:
        itinerary = Markdown(response_data["answer"])
        print(itinerary)
    else:
        # Print the response data for debugging if the answer is missing
        print("Oops, could not generate itinerary, please try again.")
        print(f"API Response: {response_data}")
